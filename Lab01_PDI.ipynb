Precisamos de fazer downsampling -> diminuir o número de pixels da imagem
Nos exemplos o número de pixels é diminuído a partir de valores discretos
Aqui o cálculo é baseado no DPI
O número de pixels já foi encontrado.

Eu sei que preciso criar uma nova matriz que vai representar a nova imagem down-sampled. Inicializo ela com a nova dimensão já encontrada.

Imagem em gray_scale tem so linha e coluna. Dessa forma, sei que preciso criar a nova imagem inicializando uma matriz vazia e passando os elementos da imagem original pra imagem nova.

1. Pegar imagem original em gray_scale (só tem linha e coluna)
2. Pegar linha e coluna da imagem original
3. Calcular dimensões da imagem nova considerando a dimensão da imagem original e o DPI da imagem nova
4. Inicializar matriz da imagem nova
5. Copiar os valores da matriz da imagem original para a matriz da imagem nova
6. Mostrar a matriz da imagem nova

```python
# dpi = pixel / polegada
# altura
  # 300 = pixel / 2.96
  # pixel = 888

# largura
  # 300 = pixel / 2.25
  # pixel = 675

# nova resolucao = (altura, largura) = (888, 675)
import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('relogio.tif', cv2.IMREAD_GRAYSCALE)

# dimensões imagem original e dpi da imagem nova
novo_dpi = 300
altura_polegada = 2.96
largura_polegada = 2.25

# cálculo das dimensões da nova imagem
nova_altura_pixel = int(novo_dpi * altura_polegada)
nova_largura_pixel = int(novo_dpi * largura_polegada)

nova_dim = (nova_altura_pixel, nova_largura_pixel)

# calculando o fator de downsampling pra saber quais pixels pegar da imagem original
downsampling_factor_altura = int(img.shape[0] / nova_altura_pixel)
downsampling_factor_largura = int(img.shape[1] / nova_largura_pixel)

nova_img = np.zeros(nova_dim,dtype=int)

nova_img = img[0:img.shape[0]:downsampling_factor_altura, 0:img.shape[1]:downsampling_factor_largura]

plt.imshow(nova_img, interpolation='bilinear', cmap='gray', vmin=0, vmax=255)
```

## Utilizando a imagem ctskull-256.tif, refaça a quantização de seus pixels utilizando de 7 a 1 bit(s) por pixel.
Fazer os cálculos por meio do seu algoritimo. Não usar o opencv para a conversão

Processo de quantização estaremos basicamente reduzindo as variações de cores da imagem por meio da redução do número de bits dentro de cada pixel.

Vemos que essa imagem tem 8 bits por canal (dtype) e que ela foi carregada em cinza, tendo então um único canal.

Dessa forma, 8 bits * 1 canal = 8 bits por pixel = 2^8 = 256 variações de cinza.

Com isso, precisamos de refazer a quantização pra diminuir de 2^8 (256) para 2^1 (2) variações por pixel.

Já entendi que devo criar uma matriz nova para receber os pixels da imagem. Porém, essa nova matriz deve comportar pixels de 2 níveis de intensidade, ou seja, 1 bit.

Como vou fazer essa redução do número de bits permitidos dentro de cada pixel?

De acordo com o livro, já vi que tenho que diminuir a intensidade para **níveis uniformemente espaçados**.

**Fatiamento por planos de bits**: um pixel possui 2^k níveis de cores, sendo k o número de bits. Cada número de bits representa uma camada de cores e pixels que, todas essas camadas somadas, formam a imagem.

A idéia deve ser refazer a quantização da imagem removendo essas camadas. Uma imagem com pixels de 7 bits é formada retirando a camada/plano do 8º bit.

Cada camada é uma imagem binária. Dessa forma, a imagem em cinza é formada pela sobreposição de múltiplas camadas binárias dessa mesma imagem, cada uma carregando um bit diferente.


```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('ctskull-256.tif', cv2.IMREAD_GRAYSCALE)

#plt.imshow(img, cmap='gray', vmin=0, vmax=255)

print(img.dtype)  # níveis de intensidade (num de bits em cada pixel)
linha = img.shape[0]
coluna = img.shape[1]
print(img.shape[0],img.shape[1]) # dimensões da imagem

nova_img = np.empty((linha,coluna), dtype= img.dtype)

# imagem de 7 bits = 2^7 = 128

# 1 2 4 8 16 32 64 128 256
# 0 1 2 3 4  5  6   7   8

# num_cores = 256 / 2^k, onde k é o número de bits removidos das camadas para refazer a quantização
# 7 bits => num_cores = 256 / 2^1 = 256 / 2 = 128
bits = 2
k = 8 - bits
passo = 2**k
num_cores = int(256 / passo)
#for k in range(1,num_cores):
#  nova_img = img // passo

nova_img[::,::] = img[::,::] // passo

quant = (2 ** bits) - 1

plt.imshow(nova_img, cmap='gray', vmin=0, vmax=quant)
```
