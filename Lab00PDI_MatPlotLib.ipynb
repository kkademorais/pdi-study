# Processamento Digital de Imagens (GBC216) - MatPlotLib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(1.0, 6.0, 0.2)

y = np.exp(x)

plt.plot(x,y,'r--')
#'b-': linha cheia na cor azul
#'r--': linha tracejada na cor vermelha
#'y*': coloca os pontos com um asterisco em amarelo
#'bo': coloca os pontos com uma bolinha azul

plt.title('Gráfico de Exponencial')

plt.text(3,300,'Gráfico de exponencial')

plt.xlabel('Valores de X')

plt.ylabel('Exponencial de X')

plt.show()
```

```python
x = np.arange(1.0,7.0, 0.2)

#y = np.exp(x)
z = np.log(x)
k = np.sin(x)
l = np.cos(x)

plt.plot(x,z,'b-',x,k,'r--',x,l,'bo')

plt.title('Gráficos Trigonométricos')

plt.xlabel('Valores de X')

plt.ylabel('Funções de X')

plt.legend(('log', 'sin', 'cos'),loc='lower right')

plt.show()
#'b-': linha cheia na cor azul
#'r--': linha tracejada na cor vermelha
#'y*': coloca os pontos com um asterisco em amarelo
#'bo': coloca os pontos com uma bolinha azul
```

```python
x = np.arange(10.0,20.0,0.1)
y = np.sin(x)
plt.subplot(1,3,1)
plt.plot(x,y,'r--')
plt.title('Sin')

k = np.cos(x)
plt.subplot(1,3,2)
plt.plot(x,k,'b--')
plt.title('Cos')

m = np.tan(x)
plt.subplot(1,3,3)
plt.plot(x,m,'g--')
plt.title('Tan')

plt.show()
```

```python
fig = plt.figure()

x = np.arange(-10.0,10.0,0.2)

y = x**2 + 4*x - 2

z = np.min(y)
p = np.argmin(y)
xp = x[p]

plt.scatter(xp,z,color='red')


plt.plot(x,y,'b--')

plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')

plt.title('Gráfico Quadrático')

plt.grid()

plt.show()
```

```python
left = [1,2,3,4]
height = [22,5,16,30]

estados= ['MG','PR','SP','RJ']

plt.bar(left,height,color='blue',tick_label=estados)

plt.title('% de Vascaínos nos Estados Brasileiros')
plt.xlabel('Estado')
plt.ylabel('Torcedores do Vasco')

plt.show()
```

```python
mu = 100
sigma = 15

x = mu + sigma * np.random.randn(10000)

plt.hist(x,bins=30,color='green')

plt.title('Histograma de QI')

plt.xlabel('QI')
plt.ylabel('Probabilidade')

plt.grid()

plt.show()
```
